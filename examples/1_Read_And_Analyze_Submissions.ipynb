### Group Submissions By Form Column and Compute Group Sizes

```python
formGroupBy = subsetDF.groupby('form')
formGps = formGroupBy.size().reset_index(name='count')
formGps = formGps.sort_values(by = ['count'])
formGps
```

### Visualize Form Statistics using Pyplot Library

```python
# Create a bar chart with form type on x axis and counts on y axis
ax = formGps.plot(x='form', y='count', kind='bar', figsize=(8,6), color='skyblue', width=0.25)
plt.title('Structured Data Form Statistics for 2026 Q2',fontsize=15)
plt.xlabel('Form Type', fontsize=15)
plt.ylabel('Number of Filings', fontsize=15)
plt.yticks(np.arange(0, 1401, step=100))
plt.xticks(rotation=0)
# Add counts on top of bars
ax.bar_label(ax.containers[0], padding=2)
plt.grid(axis='y', linestyle=':')
# Save the bar chart as a file in images folder
plt.savefig('images/FormStats2026Q2.jpg')
plt.show()
```
