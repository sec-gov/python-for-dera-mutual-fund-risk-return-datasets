<h1 align="center">Join Submissions, Numeric and Tags Data Sets to Find Custom Facts</h1>

### Import Code Libraries

```python
import pandas as pd
import numpy as np
# Set the pandas column width to unlimited
pd.set_option('display.max_colwidth', None)
```

### Read Submissions, Numeric and Tags Data Sets for from [Mutual Fund Prospectus Risk/Return Summary Data Sets](https://www.sec.gov/data-research/sec-markets-data/mutual-fund-prospectus-riskreturn-summary-data-sets)

```python
# Setting low_memory parameter to False to prevent mixed type inference and default_na parameter to False so that empty strings are not parsed as NaN.
submissions = pd.read_csv('data/2026q2_rr1/sub.tsv', sep='\t', low_memory=False, keep_default_na=False)
numericFacts = pd.read_csv('data/2026q2_rr1/num.tsv', sep='\t', low_memory=False, keep_default_na=False)
tags = pd.read_csv('data/2026q2_rr1/tag.tsv', sep='\t', low_memory=False, keep_default_na=False)
```

### Filter for Custom Monetary Tags

```python
# use value of abstarct column as 0 to filter for concrete (non-abstract) tags
customTagsDF = tags[(tags.custom == 1) & (tags.abstract == 0) & (tags.datatype == 'monetary')]
customTagsDF
```

### Merge submissions Data Frame with numericFacts Data Frame Using adsh Column as a Join Key

```python
mergedSubmissionsAndNumericFactsDF = pd.merge(submissions, numericFacts, on='adsh')
```

### Merge mergedSubmissionsAndNumericFactsDF Data Frame with customTagsDF Data Frame Using adsh and tag as Join Keys

```python
# Please note that version column = adsh for custom tags
mergedDF = pd.merge(mergedSubmissionsAndNumericFactsDF, customTagsDF, left_on=['adsh', 'tag'], right_on=['version', 'tag'])
```

### Look at the Contents of mergedDF Data Frame

```python
mergedDF
```

### Display a Subset of Columns for Custom Facts

```python
customFacts = mergedDF[['name', 'series', 'otherdims', 'tag', 'dimn', 'ddate', 'uom', 'value']].sort_values(by= ['tag', 'series', 'otherdims'])
customFacts.reset_index(drop=True, inplace=True)
customFacts
```
