<h1 align="center">Join Submission and Text Data Sets to Find Narrative Facts</h1>

### Import Code Libraries

```python
import pandas as pd
import numpy as np
```

### Read Submissions and Text Data Sets for 2026 Q2 from [Mutual Fund Prospectus Risk/Return Summary Data Sets](https://www.sec.gov/data-research/sec-markets-data/mutual-fund-prospectus-riskreturn-summary-data-sets)

```python
# Setting low_memory parameter to False to prevent mixed type inference and default_na parameter to False so that empty strings are not parsed as NaN.
submissions = pd.read_csv('data/2026q2_rr1/sub.tsv', sep='\t', low_memory=False, keep_default_na=False)
textFacts = pd.read_csv('data/2026q2_rr1/txt.tsv', sep='\t', low_memory=False, keep_default_na=False)
```

### Look at the Contents of submissions Data Frame

```python
submissions
```

### Look at the Contents of textFacts Data Frame

```python
textFacts
```

### Merge submissions Data Frame with textFacts Data Frame Using adsh Column as a Join Key

```python
mergedDF = pd.merge(submissions, textFacts, on='adsh')
```

### Look at the Contents of mergedDF Data Frame

```python
mergedDF
```

### Filter for Investment Strategy of STATE STREET INSTITUTIONAL INVESTMENT TRUST using StrategyNarrativeTextBlock tag

```python
filter = (mergedDF.name == 'STATE STREET INSTITUTIONAL INVESTMENT TRUST') & (mergedDF.tag == 'StrategyNarrativeTextBlock') 
# create a data frame with a subset of columns
textFactsForStateStreet = mergedDF[filter][['name', 'series', 'document', 'tag', 'value']]
textFactsForStateStreet = textFactsForStateStreet.sort_values(by=['series', 'document'])
# reset the index of textFactsForStateStreet data frame
textFactsForStateStreet.reset_index(drop=True, inplace=True)
#left align the columns and headers
leftAlignedDF = textFactsForStateStreet.style.set_properties(**{'text-align': 'left', 'vertical-align':'top'})
leftAlignedDF.set_table_styles([dict(selector='th', props=[('text-align', 'left')])])
```

### Filter for Investment Risks of BLACKROCK FUNDS for using RiskTextBlock

```python
filter = (mergedDF.name == 'BLACKROCK FUNDS') & (mergedDF.tag == 'RiskTextBlock') 
# create a data frame with a subset of columns
textFactsForBlackrock = mergedDF[filter][['name', 'series', 'document', 'tag', 'value']]
# reset the index of textFactsForBlackrock data frame
textFactsForBlackrock = textFactsForBlackrock.sort_values(by = ['series', 'document']).reset_index(drop=True)
#left align the columns and headers
leftAlignedDF = textFactsForBlackrock.style.set_properties(**{'text-align': 'left', 'vertical-align':'top'})
leftAlignedDF.set_table_styles([dict(selector='th', props=[('text-align', 'left')])])
```

### Create a Excel file for Risk Factors

```python
# Save the Excel file in data folder
textFactsForBlackrock.to_excel('data/RiskFactors.xlsx', index=False)
```
